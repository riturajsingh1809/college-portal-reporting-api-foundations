```python
# ==========================================================
# PlaceMux - Phase 2
# Task 16
# College Portal & Reporting API Foundations
#
# PART 1
# Import Libraries
# Load Dataset
# Dataset Exploration
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)

print("="*60)
print("College Portal Dashboard Project")
print("="*60)

# ----------------------------------------------------------
# Upload Dataset
# ----------------------------------------------------------

from google.colab import files

uploaded = files.upload()

# ----------------------------------------------------------
# Read Excel File
# ----------------------------------------------------------

df = pd.read_excel("Sample_Dataset.xlsx")

print("\nDataset Loaded Successfully.\n")

# ----------------------------------------------------------
# Display First Records
# ----------------------------------------------------------

print("First 10 Records\n")

display(df.head(10))

# ----------------------------------------------------------
# Last Records
# ----------------------------------------------------------

print("\nLast 10 Records\n")

display(df.tail(10))

# ----------------------------------------------------------
# Dataset Shape
# ----------------------------------------------------------

print("\nRows and Columns")

print(df.shape)

# ----------------------------------------------------------
# Column Names
# ----------------------------------------------------------

print("\nColumn Names\n")

print(df.columns)

# ----------------------------------------------------------
# Data Types
# ----------------------------------------------------------

print("\nDataset Information\n")

df.info()

# ----------------------------------------------------------
# Missing Values
# ----------------------------------------------------------

print("\nMissing Values\n")

print(df.isnull().sum())

# ----------------------------------------------------------
# Duplicate Values
# ----------------------------------------------------------

print("\nDuplicate Records")

print(df.duplicated().sum())

# ----------------------------------------------------------
# Statistical Summary
# ----------------------------------------------------------

print("\nStatistical Summary\n")

display(df.describe())

# ----------------------------------------------------------
# Numerical Columns
# ----------------------------------------------------------

print("\nNumerical Columns\n")

print(df.select_dtypes(include=np.number).columns)

# ----------------------------------------------------------
# Categorical Columns
# ----------------------------------------------------------

print("\nCategorical Columns\n")

print(df.select_dtypes(include="object").columns)

# ----------------------------------------------------------
# Dataset Completed
# ----------------------------------------------------------

print("\nDataset Exploration Completed Successfully.")
```

```python
# ==========================================================
# PART 2
# Data Cleaning + College Dashboard KPIs
# ==========================================================

print("="*70)
print("DATA CLEANING")
print("="*70)

# ---------------------------------------
# Missing Values
# ---------------------------------------

print("\nMissing Values\n")
print(df.isnull().sum())

# ---------------------------------------
# Duplicate Records
# ---------------------------------------

print("\nDuplicate Records")
print(df.duplicated().sum())

# Remove Duplicate Records

df = df.drop_duplicates()

print("\nDataset Shape After Removing Duplicates")
print(df.shape)

# ---------------------------------------
# Data Types
# ---------------------------------------

print("\nData Types\n")
print(df.dtypes)

# ---------------------------------------
# Unique Values
# ---------------------------------------

print("\nUnique Departments")
print(df["Department"].unique())

print("\nUnique Years")
print(df["Year"].unique())

print("\nPlacement Status")
print(df["Placement_Status"].unique())

# ---------------------------------------
# KPI SECTION
# ---------------------------------------

print("\n" + "="*70)
print("COLLEGE DASHBOARD KPI")
print("="*70)

total_students = len(df)

total_departments = df["Department"].nunique()

average_attendance = round(df["Attendance"].mean(),2)

average_marks = round(df["Marks"].mean(),2)

average_cgpa = round(df["CGPA"].mean(),2)

placed_students = (df["Placement_Status"]=="Placed").sum()

not_placed = (df["Placement_Status"]=="Not Placed").sum()

placement_percentage = round((placed_students/total_students)*100,2)

highest_marks = df["Marks"].max()

lowest_marks = df["Marks"].min()

highest_cgpa = df["CGPA"].max()

lowest_cgpa = df["CGPA"].min()

total_fees = df["Fees"].sum()

average_fees = round(df["Fees"].mean(),2)

print(f"Total Students : {total_students}")

print(f"Departments : {total_departments}")

print(f"Average Attendance : {average_attendance}")

print(f"Average Marks : {average_marks}")

print(f"Average CGPA : {average_cgpa}")

print(f"Placed Students : {placed_students}")

print(f"Not Placed Students : {not_placed}")

print(f"Placement Percentage : {placement_percentage}%")

print(f"Highest Marks : {highest_marks}")

print(f"Lowest Marks : {lowest_marks}")

print(f"Highest CGPA : {highest_cgpa}")

print(f"Lowest CGPA : {lowest_cgpa}")

print(f"Total Fees Collection : {total_fees}")

print(f"Average Fees : {average_fees}")

# ---------------------------------------
# Department Wise Student Count
# ---------------------------------------

print("\nDepartment Wise Students\n")

department_summary = df.groupby("Department")["Student_ID"].count()

print(department_summary)

# ---------------------------------------
# Year Wise Students
# ---------------------------------------

print("\nYear Wise Students\n")

year_summary = df.groupby("Year")["Student_ID"].count()

print(year_summary)

# ---------------------------------------
# Gender Distribution
# ---------------------------------------

print("\nGender Distribution\n")

gender_summary = df["Gender"].value_counts()

print(gender_summary)

print("\nPART 2 COMPLETED SUCCESSFULLY")
```

```python
# ==========================================================
# PART 3
# College Dashboard Charts & Visualization
# ==========================================================

import matplotlib.pyplot as plt

print("="*70)
print("COLLEGE DASHBOARD VISUALIZATION")
print("="*70)

# ---------------------------------------------------------
# Chart 1 : Department Wise Students
# ---------------------------------------------------------

dept = df["Department"].value_counts()

plt.figure(figsize=(8,5))
dept.plot(kind="bar")

plt.title("Department Wise Students")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Chart 2 : Gender Distribution
# ---------------------------------------------------------

gender = df["Gender"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender,
    labels=gender.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")

plt.show()

# ---------------------------------------------------------
# Chart 3 : Placement Status
# ---------------------------------------------------------

placement = df["Placement_Status"].value_counts()

plt.figure(figsize=(6,5))

placement.plot(kind="bar")

plt.title("Placement Status")

plt.xlabel("Status")

plt.ylabel("Students")

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 4 : Attendance Distribution
# ---------------------------------------------------------

plt.figure(figsize=(8,5))

plt.hist(df["Attendance"], bins=10)

plt.title("Attendance Distribution")

plt.xlabel("Attendance")

plt.ylabel("Students")

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 5 : Marks Distribution
# ---------------------------------------------------------

plt.figure(figsize=(8,5))

plt.hist(df["Marks"], bins=10)

plt.title("Marks Distribution")

plt.xlabel("Marks")

plt.ylabel("Students")

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 6 : CGPA Distribution
# ---------------------------------------------------------

plt.figure(figsize=(8,5))

plt.hist(df["CGPA"], bins=10)

plt.title("CGPA Distribution")

plt.xlabel("CGPA")

plt.ylabel("Students")

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 7 : Year Wise Students
# ---------------------------------------------------------

year = df["Year"].value_counts().sort_index()

plt.figure(figsize=(7,5))

plt.plot(year.index, year.values, marker="o")

plt.title("Year Wise Students")

plt.xlabel("Year")

plt.ylabel("Students")

plt.grid(True)

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 8 : Average Marks by Department
# ---------------------------------------------------------

avg_marks = df.groupby("Department")["Marks"].mean()

plt.figure(figsize=(8,5))

avg_marks.plot(kind="bar")

plt.title("Average Marks by Department")

plt.xlabel("Department")

plt.ylabel("Average Marks")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 9 : Average CGPA by Department
# ---------------------------------------------------------

avg_cgpa = df.groupby("Department")["CGPA"].mean()

plt.figure(figsize=(8,5))

avg_cgpa.plot(kind="bar")

plt.title("Average CGPA by Department")

plt.xlabel("Department")

plt.ylabel("Average CGPA")

plt.xticks(rotation=20)

plt.tight_layout()

plt.show()

# ---------------------------------------------------------
# Chart 10 : Fees Distribution
# ---------------------------------------------------------

plt.figure(figsize=(8,5))

plt.boxplot(df["Fees"])

plt.title("Fees Distribution")

plt.ylabel("Fees")

plt.tight_layout()

plt.show()

print("="*70)
print("PART 3 COMPLETED SUCCESSFULLY")
print("="*70)
```

```python
# ==========================================================
# PART 4
# Executive Dashboard Report & Final Excel Report
# ==========================================================

print("="*80)
print("EXECUTIVE DASHBOARD REPORT")
print("="*80)

# ----------------------------------------------------------
# Overall Summary
# ----------------------------------------------------------

summary = pd.DataFrame({

    "Metric":[
        "Total Students",
        "Departments",
        "Average Attendance",
        "Average Marks",
        "Average CGPA",
        "Placed Students",
        "Not Placed Students",
        "Placement Percentage",
        "Highest Marks",
        "Highest CGPA",
        "Total Fees Collection"
    ],

    "Value":[
        len(df),
        df["Department"].nunique(),
        round(df["Attendance"].mean(),2),
        round(df["Marks"].mean(),2),
        round(df["CGPA"].mean(),2),
        (df["Placement_Status"]=="Placed").sum(),
        (df["Placement_Status"]=="Not Placed").sum(),
        round(((df["Placement_Status"]=="Placed").sum()/len(df))*100,2),
        df["Marks"].max(),
        df["CGPA"].max(),
        df["Fees"].sum()
    ]

})

print("\nOVERALL SUMMARY\n")

display(summary)

# ----------------------------------------------------------
# Department Report
# ----------------------------------------------------------

department_report = df.groupby("Department").agg({

    "Student_ID":"count",
    "Attendance":"mean",
    "Marks":"mean",
    "CGPA":"mean",
    "Fees":"sum"

}).round(2)

department_report.columns=[
    "Students",
    "Average Attendance",
    "Average Marks",
    "Average CGPA",
    "Total Fees"
]

print("\nDEPARTMENT REPORT\n")

display(department_report)

# ----------------------------------------------------------
# Placement Report
# ----------------------------------------------------------

placement_report=df.groupby("Placement_Status").agg({

    "Student_ID":"count",
    "Marks":"mean",
    "CGPA":"mean"

}).round(2)

placement_report.columns=[
    "Students",
    "Average Marks",
    "Average CGPA"
]

print("\nPLACEMENT REPORT\n")

display(placement_report)

# ----------------------------------------------------------
# Gender Report
# ----------------------------------------------------------

gender_report=df.groupby("Gender").agg({

    "Student_ID":"count",
    "Attendance":"mean",
    "Marks":"mean",
    "CGPA":"mean"

}).round(2)

gender_report.columns=[
    "Students",
    "Average Attendance",
    "Average Marks",
    "Average CGPA"
]

print("\nGENDER REPORT\n")

display(gender_report)

# ----------------------------------------------------------
# Top 10 Students by Marks
# ----------------------------------------------------------

top_marks=df.sort_values("Marks",ascending=False)

print("\nTOP 10 STUDENTS BY MARKS\n")

display(top_marks.head(10))

# ----------------------------------------------------------
# Top 10 Students by CGPA
# ----------------------------------------------------------

top_cgpa=df.sort_values("CGPA",ascending=False)

print("\nTOP 10 STUDENTS BY CGPA\n")

display(top_cgpa.head(10))

# ----------------------------------------------------------
# Save Reports to Excel
# ----------------------------------------------------------

with pd.ExcelWriter("College_Final_Report.xlsx") as writer:

    df.to_excel(writer,
                sheet_name="Complete Dataset",
                index=False)

    summary.to_excel(writer,
                     sheet_name="Summary",
                     index=False)

    department_report.to_excel(writer,
                               sheet_name="Department Report")

    placement_report.to_excel(writer,
                              sheet_name="Placement Report")

    gender_report.to_excel(writer,
                           sheet_name="Gender Report")

    top_marks.head(10).to_excel(writer,
                                sheet_name="Top Marks",
                                index=False)

    top_cgpa.head(10).to_excel(writer,
                               sheet_name="Top CGPA",
                               index=False)

print("\nCollege_Final_Report.xlsx Saved Successfully")

print("="*80)
print("PART 4 COMPLETED SUCCESSFULLY")
print("="*80)
```

```python
# ==========================================================
# PART 5
# Final Reports Export & Project Completion
# ==========================================================

import json

print("="*80)
print("FINAL PROJECT REPORT")
print("="*80)

# ----------------------------------------------------------
# Final Summary Dictionary
# ----------------------------------------------------------

summary = {

    "Total Students": int(len(df)),
    "Departments": int(df["Department"].nunique()),
    "Average Attendance": round(df["Attendance"].mean(),2),
    "Average Marks": round(df["Marks"].mean(),2),
    "Average CGPA": round(df["CGPA"].mean(),2),
    "Placed Students": int((df["Placement_Status"]=="Placed").sum()),
    "Not Placed Students": int((df["Placement_Status"]=="Not Placed").sum()),
    "Placement Percentage": round(((df["Placement_Status"]=="Placed").sum()/len(df))*100,2),
    "Highest Marks": int(df["Marks"].max()),
    "Highest CGPA": float(df["CGPA"].max()),
    "Total Fees Collection": int(df["Fees"].sum())
}

# ----------------------------------------------------------
# Save JSON Report
# ----------------------------------------------------------

with open("College_Report.json","w") as file:
    json.dump(summary,file,indent=4)

print("College_Report.json Saved Successfully")

# ----------------------------------------------------------
# Save CSV Dataset
# ----------------------------------------------------------

df.to_csv("College_Dataset.csv",index=False)

print("College_Dataset.csv Saved Successfully")

# ----------------------------------------------------------
# Save Department Report
# ----------------------------------------------------------

department_report = df.groupby("Department").agg({
    "Student_ID":"count",
    "Attendance":"mean",
    "Marks":"mean",
    "CGPA":"mean"
}).round(2)

department_report.to_csv("Department_Report.csv")

print("Department_Report.csv Saved Successfully")

# ----------------------------------------------------------
# Save Placement Report
# ----------------------------------------------------------

placement_report = df.groupby("Placement_Status").agg({
    "Student_ID":"count",
    "Marks":"mean",
    "CGPA":"mean"
}).round(2)

placement_report.to_csv("Placement_Report.csv")

print("Placement_Report.csv Saved Successfully")

# ----------------------------------------------------------
# Print Final Summary
# ----------------------------------------------------------

print("\n========== PROJECT SUMMARY ==========\n")

for key,value in summary.items():
    print(f"{key} : {value}")

print("\n=====================================")
print("PROJECT COMPLETED SUCCESSFULLY")
print("=====================================")

print("\nGenerated Files:")
print("1. Sample_Dataset.xlsx")
print("2. College_Final_Report.xlsx")
print("3. College_Dataset.csv")
print("4. Department_Report.csv")
print("5. Placement_Report.csv")
print("6. College_Report.json")

print("\nThank You!")
```
